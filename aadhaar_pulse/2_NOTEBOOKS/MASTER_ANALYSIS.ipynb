# 🔬 MODULE 1: Migration Corridor Detection

**Key Insight**: High demographic updates + Low enrolments = People MOVING IN

```python
# Calculate migration indicators
migration_state = calculate_migration_indicators(enrol_df, demo_df, level='state')
migration_district = calculate_migration_indicators(enrol_df, demo_df, level='district')

# Print summary
summary = generate_migration_summary(enrol_df, demo_df)
print_migration_report(summary)
```

```python
# Visualize migration patterns
fig, ax = plot_top_n_bar(migration_state, 'state', 'migration_score', 
                         'Top States by Migration Activity Score', n=15)
plt.savefig('../5_OUTPUTS/charts/migration_by_state.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Migration type breakdown
migration_types = migration_state['migration_type'].value_counts()
fig, ax = plt.subplots(figsize=(10, 8))
colors = ['#2E86AB', '#A23B72', '#F18F01', '#C73E1D', '#457B9D']
ax.pie(migration_types.values, labels=migration_types.index, autopct='%1.1f%%', 
       colors=colors[:len(migration_types)], explode=[0.02]*len(migration_types))
ax.set_title('States by Migration Type', fontsize=16, fontweight='bold')
plt.savefig('../5_OUTPUTS/charts/migration_types_pie.png', dpi=150, bbox_inches='tight')
plt.show()
```

# 👻 MODULE 2: Anomaly Detection (Ghost Districts)

**Key Insight**: Districts with high enrolments but zero updates = potential fraud

```python
# Run anomaly detection
anomaly_df = calculate_anomaly_scores(enrol_df, demo_df, bio_df, level='district')
ghost_districts = detect_ghost_districts(enrol_df, demo_df, bio_df, top_n=20)

# Print summary
anomaly_summary = generate_anomaly_summary(enrol_df, demo_df, bio_df)
print_anomaly_report(anomaly_summary)
```

```python
# Visualize anomaly distribution
fig, ax = plot_distribution(anomaly_df['anomaly_score'], 
                            'Distribution of Anomaly Scores Across Districts',
                            'Anomaly Score')
plt.savefig('../5_OUTPUTS/charts/anomaly_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Top ghost districts
if len(ghost_districts) > 0:
    ghost_districts['label'] = ghost_districts['district'] + ', ' + ghost_districts['state']
    fig, ax = plot_top_n_bar(ghost_districts, 'label', 'risk_score',
                             '🚨 Top Ghost Districts by Risk Score', n=10)
    plt.savefig('../5_OUTPUTS/charts/ghost_districts.png', dpi=150, bbox_inches='tight')
    plt.show()
```

# 📊 MODULE 3: Predictive Service Demand

**Key Insight**: 0-5 enrolments today = School seats needed in 5-13 years

```python
# Generate forecasts
school_forecast = forecast_school_demand(enrol_df, years_ahead=5)
healthcare_forecast = forecast_healthcare_demand(enrol_df, bio_df)
pension_forecast = forecast_pension_beneficiaries(enrol_df, demo_df, years_ahead=10)

# Print summary
demand_summary = generate_demand_forecast_summary(enrol_df, demo_df, bio_df)
print_demand_forecast_report(demand_summary)
```

# 🌊 MODULE 6: Crisis Early Warning System

**Key Insight**: Detect disasters 24-48 hours before official reports

```python
# Detect crisis signals
crisis_df = detect_crisis_signals(enrol_df, demo_df, bio_df)
crisis_summary = get_crisis_summary(enrol_df, demo_df, bio_df)

print("\n" + "="*60)
print("🚨 CRISIS EARLY WARNING SYSTEM")
print("="*60)
print(f"\nTotal Days Analyzed: {crisis_summary['total_days']}")
print(f"Crisis Days Detected: {crisis_summary['crisis_days']}")
print(f"\nAlert Breakdown:")
for level, count in crisis_summary['alert_breakdown'].items():
    emoji = {'RED': '🔴', 'ORANGE': '🟠', 'YELLOW': '🟡', 'GREEN': '🟢', 'DEAD_ZONE': '⚫'}.get(level, '⚪')
    print(f"   {emoji} {level}: {count} days")
```

```python
# Visualize crisis timeline
fig, ax = plt.subplots(figsize=(16, 6))
ax.plot(crisis_df['date'], crisis_df['total_activity'], color='#2E86AB', alpha=0.7)
ax.fill_between(crisis_df['date'], crisis_df['total_activity'], alpha=0.3)

# Highlight crisis days
crisis_days = crisis_df[crisis_df['is_crisis'] == 1]
ax.scatter(crisis_days['date'], crisis_days['total_activity'], color='red', s=50, zorder=5)

ax.set_title('🌊 Crisis Early Warning Timeline', fontsize=16, fontweight='bold')
ax.set_xlabel('Date')
ax.set_ylabel('Total Activity')
plt.savefig('../5_OUTPUTS/charts/crisis_timeline.png', dpi=150, bbox_inches='tight')
plt.show()
```

# 📈 MODULE 7: Aadhaar Sentiment Index

**Key Insight**: Measure national well-being without costly surveys

```python
# Calculate sentiment index
sentiment_df = calculate_sentiment_index(enrol_df, demo_df, bio_df, level='state')

print("\n" + "="*60)
print("📈 AADHAAR SENTIMENT INDEX")
print("="*60)
print(f"\nNational Average: {sentiment_df['sentiment_index'].mean():.1f}/100")
print(f"\nSentiment by Category:")
for cat, count in sentiment_df['sentiment_category'].value_counts().items():
    emoji = {'THRIVING': '🟢', 'STABLE': '🟡', 'STRUGGLING': '🟠', 'AT_RISK': '🔴'}.get(cat, '⚪')
    print(f"   {emoji} {cat}: {count} states")
```

```python
# Visualize sentiment gauge (national average)
national_sentiment = sentiment_df['sentiment_index'].mean() / 100
fig, ax = plot_sentiment_gauge(national_sentiment, 'National Aadhaar Sentiment Index')
plt.savefig('../5_OUTPUTS/charts/sentiment_gauge.png', dpi=150, bbox_inches='tight')
plt.show()
```

# 👶 MODULE 8: Cohort Tracking

**Key Insight**: Track children enrolled at 0-5 who never appear in subsequent updates

```python
# Track cohort progression
cohort_data = track_cohort_progression(enrol_df, demo_df, bio_df)

print("\n" + "="*60)
print("👶 COHORT TRACKING ANALYSIS")
print("="*60)
print(f"\nEnrolled:")
print(f"   Age 0-5: {cohort_data['enrolled']['age_0_5_enrolled']:,}")
print(f"   Age 5-17: {cohort_data['enrolled']['age_5_17_enrolled']:,}")
print(f"   Age 18+: {cohort_data['enrolled']['age_18_plus_enrolled']:,}")
print(f"\nSurvival Rates:")
print(f"   0-5 → 5-17: {cohort_data['survival_rate_0_5_to_5_17']*100:.1f}%")
print(f"   5-17 → Adult: {cohort_data['survival_rate_5_17_to_adult']*100:.1f}%")
print(f"\n⚠️ At-Risk Children: {cohort_data['at_risk_children']:,.0f}")
```

```python
# Visualize cohort funnel
cohort_funnel = {
    'Enrolled (0-5)': cohort_data['enrolled']['age_0_5_enrolled'],
    'Demo Updated (5-17)': cohort_data['demo_updated']['age_5_17_demo_updated'],
    'Bio Updated (5-17)': cohort_data['bio_updated']['age_5_17_bio_updated']
}
fig, ax = plot_cohort_funnel(cohort_funnel, 'Child Cohort Progression Funnel')
plt.savefig('../5_OUTPUTS/charts/cohort_funnel.png', dpi=150, bbox_inches='tight')
plt.show()
```
